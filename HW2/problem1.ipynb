```python
import numpy as np

def getEigen(arr):
    eigenvalues, eigenvectors = np.linalg.eig(arr)

    for i, value in enumerate(eigenvalues):
        print(f"Eigenvalue {i + 1}: {value:.3f}")
        print(f"Eigenvector {i + 1}: {np.round(eigenvectors[:, i], 3)}")
        print()
```

### Problem 1A

```python
arr = np.zeros((10,10))

for i in range(len(arr)):
    arr[i][i] = 1

print(arr)
```

```python
getEigen(arr)
```

### Problem 1B

```python
rank = np.linalg.matrix_rank(arr)
print(f"Rank: {rank}")
```

### 1C

```python
random = np.random.randint(0, 2, size=(10, 10))
print(random)
```

```python
new_arr = arr + random
print(new_arr)
```

```python
getEigen(new_arr)
```

```python
rank = np.linalg.matrix_rank(new_arr)
print(f"Rank: {rank}")
```
