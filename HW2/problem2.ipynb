### 2A

```python
import numpy as np
from numpy.polynomial import Polynomial
import matplotlib.pyplot as plt

coefficients = [1, -18, 144, -672, 2016, -4032, 5376, -4608, 2304, -512]
reversed_coefficients = coefficients[::-1]

p = Polynomial(reversed_coefficients)

x = np.linspace(1.920, 2.080, 161)

y1 = p(x)

plt.plot(x, y1, label="p(x)")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid()

plt.show()
```

### 2B

```python
y2 = (x - 2)**9
plt.plot(x, y2, label="(x - 2)^9")

plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid()

plt.show()
```

### 2C

```python
plt.plot(x, y1, label="p(x)")
plt.plot(x, y2, label="(x - 2)^9")

plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid()

plt.show()
```

Despite being equivalent, they appear different on graph since p(x) is more affected by loss of numeric precision due to floating point arithmetic.
